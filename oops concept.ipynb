```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute (prefixed with double underscore)

    def deposit(self, amount):
        self.__balance += amount

    def display_balance(self):
        print("balance:", self.__balance)


# Example usage:
account = BankAccount(1000)
account.deposit(500)
account.display_balance()  # Output: balance: 1500
```

```python
#poly = same method different behave
#inharitance = use parent
#single inheritance
class BankAccount:
    def __init__(self, name, age, balance):#name,age,balnce are parameter
        self.name = name
        self.age = age
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

#child class
class SavingAccount(BankAccount):
    def display_balance(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Balance:", self.balance)


account = SavingAccount("Yadhu", 21, 5000)

account.deposit(2000)
account.display_balance()
```

```python
# multiple inheritance
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount


class BankSecurity:
    def security(self):
        print("Account is verified")


class SavingAccount(BankAccount, BankSecurity):
    def show_balance(self):
        print("Balance is:", self.balance)


account = SavingAccount("Yadhu", 5000)

account.deposit(2000)
account.security()
account.show_balance()
```

```python
#multi level inheritance
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount


class SavingAccount(BankAccount):
    def savings(self):
        print("Saving account")


class PremiumSavingAccount(SavingAccount):
    def display(self):
        print("Name:", self.name)
        print("Balance:", self.balance)


account = PremiumSavingAccount("Yadhu", 5000)

account.deposit(2000)
account.savings()
account.display()
        
```
